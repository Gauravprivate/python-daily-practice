```python
import threading
import time

def download_file(filename):
    print(f"Downloading {filename}...")
    time.sleep(2)
    print(f"{filename} downloaded successfully!")

thread1 = threading.Thread(target=download_file, args=("File1.pdf",))
thread2 = threading.Thread(target=download_file, args=("File2.pdf",))
thread3 = threading.Thread(target=download_file, args=("File3.pdf",))

thread1.start()
thread2.start()
thread3.start()

thread1.join()
thread2.join()
thread3.join()

print("All files downloaded!")
```

```python
import multiprocessing
import time

def process_file(filename):
    print(f"Processing {filename}...")
    time.sleep(2)
    print(f"{filename} processed successfully!")

if __name__ == "__main__":
    process1 = multiprocessing.Process(
        target=process_file,
        args=("Sales.csv",)
    )

    process2 = multiprocessing.Process(
        target=process_file,
        args=("Employees.csv",)
    )

    process1.start()
    process2.start()

    process1.join()
    process2.join()

    print("All files processed!")
```

```python
import asyncio

async def fetch_customer():
    print("Fetching customer data...")
    await asyncio.sleep(2)
    print("Customer data loaded!")

async def fetch_products():
    print("Fetching product data...")
    await asyncio.sleep(3)
    print("Product data loaded!")

async def fetch_orders():
    print("Fetching order data...")
    await asyncio.sleep(1)
    print("Order data loaded!")

async def main():
    await asyncio.gather(
        fetch_customer(),
        fetch_products(),
        fetch_orders()
    )

await main()

print("Dashboard loaded successfully!")
```

```python
import asyncio

async def check_stock():
    print("Checking product stock...")
    await asyncio.sleep(2)
    print("Product is in stock!")

async def process_payment():
    print("Processing payment...")
    await asyncio.sleep(3)
    print("Payment successful!")

async def send_confirmation():
    print("Sending order confirmation...")
    await asyncio.sleep(1)
    print("Confirmation email sent!")

async def main():
    await asyncio.gather(
        check_stock(),
        process_payment(),
        send_confirmation()
    )

await main()

print("Order processing completed!")
```
