```python
x=int(input("Enter First Number"))
y=int(input("Enter Second Number"))
z=int(input("Enter Third Number"))

if x > y and x > z:
    print("first number is greater",x)
elif y > x and y > z:
    print("second number is greater",y)
elif z > x and z > y:
    print("third number is greater",z)
else:
    print("there is Tie")
```
