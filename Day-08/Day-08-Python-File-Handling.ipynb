```python
file = open("about.txt","w")
file.write("My name is Gaurav.\n")
file.write("I am learning python.\n")
file.write("I want to become a python Developer.\n")
file.close()

file = open("about.txt","r")
content = file.read()
print(content)
file.close()
```

```python
file = open("about.txt","w")
file.write("My name is Gaurav.\n")
file.write("I am learning python.\n")
file.write("I want to become a python Developer.\n")
file.close()

with open("about.txt","r")as file:
    content = file.read()
    print(content)
```

```python
file = open("about.txt","w")
file.write("My name is Gaurav.\n")
file.write("I am learning python.\n")
file.write("I want to become a python Developer.\n")
file.close()

with open("about.txt","a")as file:
    file.write("I practice python everyday.\n")
    file.write("I am building my GitHub portfolio.\n")

with open("about.txt","r")as file:
    content = file.read()
    print(content)
```

```python
with open("about.txt", "r") as file:
    lines = file.readlines()

print("Number of lines:", len(lines))
```

```python
with open("about.txt", "r") as file:
    content = file.read()

words = content.split()
print("Number of words:", len(words))
```

```python
with open("about.txt", "r") as file:
    content = file.read()

words = content.lower().split()
count = words.count("python")

print("Python appears", count, "times.")
```

```python
message = input("Enter a sentence: ")

with open("message.txt", "w") as file:
    file.write(message)

with open("message.txt", "r") as file:
    content = file.read()

print("Saved message:", content)

words = content.split()
print("Word count:", len(words))
```
