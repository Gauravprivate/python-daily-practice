```python
import re

text = "Contact us at support@gmail.com or hr@company.in"

emails = re.findall(r"[\w.]+@[\w.]+", text)

print(emails)
```

```python
import re

text = "My numbers are 9876543210 and 8765432109"

result = re.findall(r"\d{10}", text)

print(result)
```

```python
import re

phone1 = "9876543210"
phone2 = "98765432"
phone3 = "98765abc10"

phone_numbers = [phone1, phone2, phone3]

for phone in phone_numbers:
    if re.fullmatch(r"\d{10}", phone):
        print(phone, "- Valid")
    else:
        print(phone, "- Invalid")
```
