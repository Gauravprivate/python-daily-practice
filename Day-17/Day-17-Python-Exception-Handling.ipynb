```python
# Real Word Example
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Invalid withdrawal amount!")

        if amount > self.__balance:
            raise Exception("Insufficient balance!")

        self.__balance -= amount
        print("Withdrawal successful!")


account1 = BankAccount("Gaurav", 10000)

try:
    account1.withdraw(2000)
    account1.withdraw(15000)
    account1.withdraw(-500)

except ValueError as e:
    print(e)

except Exception as e:
    print(e)
```

```python

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def validate_marks(self):
        if self.marks < 0:
            raise ValueError("Marks cannot be negative!")

        elif self.marks > 100:
            raise ValueError("Marks cannot exceed 100!")

        else:
            print("Marks accepted successfully!")


student1 = Student("Nikhil", 85)
student2 = Student("Pranav", -10)
student3 = Student("Durgesh", 120)

for student in [student1, student2, student3]:
    try:
        student.validate_marks()

    except ValueError as e:
        print(student.name, "-", e)
```
