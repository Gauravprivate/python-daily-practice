```python
import threading
import time

def task():
    print("Task started")
    time.sleep(2)
    print("Task completed")

thread1 = threading.Thread(target=task)
thread2 = threading.Thread(target=task)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("All tasks completed")
```

```python
import threading
import time

def download_file():
    print("Downloading file...")
    time.sleep(2)
    print("Download completed!")

thread1 = threading.Thread(target=download_file)
thread2 = threading.Thread(target=download_file)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("All downloads finished!")
```

```python
# Real word Example

import threading
import time

def send_email():
    print("Sending confirmation email...")
    time.sleep(2)
    print("Email sent!")

def send_sms():
    print("Sending SMS notification...")
    time.sleep(2)
    print("SMS sent!")

thread1 = threading.Thread(target=send_email)
thread2 = threading.Thread(target=send_sms)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("Order notifications completed!")
```

```python
import multiprocessing
import time

def process_data(name):
    print("Processing started:", name)
    time.sleep(2)
    print("Processing completed:", name)


if __name__ == "__main__":

    process1 = multiprocessing.Process(
        target=process_data,
        args=("Process 1",)
    )

    process2 = multiprocessing.Process(
        target=process_data,
        args=("Process 2",)
    )

    process1.start()
    process2.start()

    process1.join()
    process2.join()

    print("All data processing completed!")
```
