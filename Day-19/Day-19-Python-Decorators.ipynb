```python
def my_decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper


@my_decorator
def say_hello():
    print("Hello!")


say_hello()
```

```python
def my_decorator(func):
    def greet():
        print("Starting Function")
        func()
        print("Function Completed")
    return greet

@my_decorator
def welcome():
    print("Welcome to Python!")

welcome()
```

```python
def login_logger(func):
    def login():
        print("User Login Started")
        func()
        print("User Login Completed")
    return login

@login_logger
def welcome():
    print("Employee Logged in")

welcome()
```

```python
def message_decorator(func):
    def wrapper(name):
        print("Function is starting")
        func(name)
        print("Function has Finished")
    return wrapper

@message_decorator
def show_user(name):
    print("Welcome", name)


show_user("Gaurav")
```

```python
def time_logger(func):
    def wrapper():
        print("Task Started")
        func()
        print("Task Completed")
    return wrapper

@time_logger
def process_data():
    for i in range(1000000):
        pass
    print("Data processed")

process_data()
```

```python
import time

def time_logger(func):
    def wrapper():
        print("Task Started")
        start = time.time()
        func()
        end = time.time()
        print("Task Completed")
        print("Execution Time:", end - start, "seconds")
    return wrapper


@time_logger
def process_data():
    for i in range(1000000):
        pass

    print("Data processed")

process_data()
```
