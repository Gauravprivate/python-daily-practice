```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s"
)

def place_order(customer, product):
    logging.info(f"{customer} placed an order for {product}")

    logging.info("Checking product availability")

    logging.info("Order placed successfully")

place_order("Gaurav", "Laptop")
```

```python
import logging

logging.basicConfig(
    level=logging.ERROR,
    format="%(levelname)s: %(message)s"
)

def process_payment(amount):
    try:
        if amount <= 0:
            raise ValueError("Payment amount must be greater than zero")

        print("Payment successful!")

    except ValueError:
        logging.exception("Payment processing failed")

process_payment(-500)
```

```python
import logging

logging.basicConfig(
    filename="login.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

def login(username, password):
    if username == "admin" and password == "1234":
        logging.info(f"{username} logged in successfully")
        print("Login successful!")

    else:
        logging.warning(f"Failed login attempt for {username}")
        print("Invalid username or password")

login("admin", "1234")
login("gaurav", "wrong")
```
