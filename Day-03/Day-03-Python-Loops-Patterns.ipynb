```python
i = 1

while i <= 10:
    print(i)
    i += 1
```

```python
i = 1

while i <= 10:
    if i == 5:
        break
    print(i)
    i += 1
```

```python
i = 1
while i <= 10:
  if i == 5:
    i += 1
    continue  
  print(i, end=" ")
  i += 1
```

```python
for i in range(1,6):
    for j in range(i):
        print("*",end="")
    print()
```

```python
for i in range(5,0,-1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
for i in range(1,6):
    for j in range(1,i + 1):
        print(j,end="")
    print()
```
