```python
def square(num):
    print(num * num)
square(5)
```

```python
def add(num1,num2):
    print(num1 + num2)
add(10,20)
```

```python
def multiply(num1 , num2):
    return num1 * num2
result = multiply(4,5)
print(result)
```

```python
def is_even(num):
    if num % 2 == 0:
        return True
    else:
        return False

print(is_even(4))
print(is_even(7))
```
