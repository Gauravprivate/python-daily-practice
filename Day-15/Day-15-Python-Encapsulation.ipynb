```python
class Student:
    def __init__(self , name , marks):
        self.__name = name
        self.__marks = marks

    def show_details(self):
        print("Name:",self.__name)
        print("Marks:",self.__marks)

student1 = Student("Gaurav", 85)

student1.show_details()
```

```python
class BankAccount:
    def __init__(self,owner , balance):
        self.owner = owner
        self.__balance = balance


    def deposit(self,amount):
        self.__balance += amount

    def show_balance(self):
        print("Account Holder:", self.owner)
        print("Balance:",self.__balance)

bankaccount1 = BankAccount("Gaurav",5000)
bankaccount1.deposit(1500)
bankaccount1.show_balance()
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.__salary = salary


    def increase_salary(self,amount):
        self.__salary += amount


    def show_details(self):
        print("Employee Name:",self.name)
        print("Salary:",self.__salary)

employee1 = Employee("Gaurav",25000)
employee1.increase_salary(5000)
employee1.show_details()
```

```python
# Real word Example
class Wallet:
    def __init__(self, customer_name, balance):
        self.customer_name = customer_name
        self.__balance = balance

    def add_money(self, amount):
        if amount <= 0:
            print("Invalid deposit amount!")
        else:
            self.__balance += amount
            print("Money added successfully:", amount)

    def make_payment(self, amount):
        if amount <= 0:
            print("Invalid payment amount!")
        elif amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            print("Payment successful:", amount)

    def check_balance(self):
        print("Customer:", self.customer_name)
        print("Available Balance:", self.__balance)


wallet1 = Wallet("Omkar", 5000)

wallet1.check_balance()
wallet1.add_money(2000)
wallet1.make_payment(1500)
wallet1.make_payment(7000)
wallet1.add_money(-500)
wallet1.check_balance()
```
