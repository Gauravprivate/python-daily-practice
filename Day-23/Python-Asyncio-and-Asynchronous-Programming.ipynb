```python
import asyncio

async def task1():
    print("Task 1 started")
    await asyncio.sleep(2)
    print("Task 1 completed")

async def task2():
    print("Task 2 started")
    await asyncio.sleep(2)
    print("Task 2 completed")

async def main():
    await asyncio.gather(task1(), task2())

await main()

print("All tasks completed")
```

```python
import asyncio

async def check_product():
    print("Checking product availability...")
    await asyncio.sleep(2)
    print("Product is available!")

async def check_delivery():
    print("Checking delivery status...")
    await asyncio.sleep(2)
    print("Delivery is on the way!")

async def main():
    await asyncio.gather(
        check_product(),
        check_delivery()
    )

await main()

print("All checks completed!")
```

```python
import asyncio

async def verify_payment():
    print("Verifying payment...")
    await asyncio.sleep(2)
    print("Payment verified!")

async def update_order():
    print("Updating order status...")
    await asyncio.sleep(2)
    print("Order status updated!")

async def send_notification():
    print("Sending confirmation notification...")
    await asyncio.sleep(2)
    print("Notification sent!")

async def main():
    await asyncio.gather(
        verify_payment(),
        update_order(),
        send_notification()
    )

await main()

print("Order processing completed!")
```

```python
import asyncio

async def fetch_user():
    print("Fetching user information...")
    await asyncio.sleep(2)
    print("User: Gaurav")

async def fetch_products():
    print("Fetching product information...")
    await asyncio.sleep(3)
    print("Products: Laptop, Mobile, Headphones")

async def fetch_orders():
    print("Fetching order information...")
    await asyncio.sleep(1)
    print("Orders: 3 orders found")

async def main():
    await asyncio.gather(
        fetch_user(),
        fetch_products(),
        fetch_orders()
    )

await main()

print("Dashboard data loaded successfully!")
```

```python
import asyncio

async def prepare_food():
    print("Restaurant is preparing food...")
    await asyncio.sleep(3)
    print("Food is ready!")

async def pickup_order():
    print("Delivery partner is arriving...")
    await asyncio.sleep(2)
    print("Order picked up!")

async def update_customer():
    print("Sending order update...")
    await asyncio.sleep(1)
    print("Customer notified!")

async def main():
    await asyncio.gather(
        prepare_food(),
        pickup_order(),
        update_customer()
    )

await main()

print("Food delivery process completed!")
```
