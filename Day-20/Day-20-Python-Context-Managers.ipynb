```python
class Demo:
    def __enter__(self):
        print("Entering the with block")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Leaving the with block")


with Demo():
    print("Inside the with block")
```

```python
class DatabaseConnection:
    def __enter__(self):
        print("Database Connected")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Database Disconnected")


with DatabaseConnection():
    print("Executing query")
```

```python
class FileManager:
    def __enter__(self):
        print("File Opened")
        return self

    def __exit__(self,exc_type,exc_value,traceback):
        print("File Closed")

with FileManager():
    print("Writing Data to  file")
```

```python
class FileManager:
    def __enter__(self):
        print("File Opened")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("File Closed")


with FileManager() as file:
    print("Writing data to file")
    print(file)
```

```python
class DatabaseManager:
    def __enter__(self):
        print("Connecting to Database...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("Closing Databse Connection...")


with DatabaseManager()as db:
    db.execute_query()
```
