```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class UPI(Payment):
    def pay(self, amount):
        print("Paid using UPI:", amount)


payment1 = UPI()
payment1.pay(500)
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class UPIPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class CardPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using Card")


upi = UPIPayment()
card = CardPayment()

upi.pay(1200)
card.pay(2500)
```

```python
from abc import ABC, abstractmethod

class Notification(ABC):

    @abstractmethod
    def send(self, message):
        pass


class EmailNotification(Notification):
    def send(self, message):
        print("Email:", message)


class SMSNotification(Notification):
    def send(self, message):
        print("SMS:", message)


email = EmailNotification()
sms = SMSNotification()

email.send("Your order has been shipped!")
sms.send("Your OTP is 123456")
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):
    def start(self):
        print("Car starts using a key")


class Bike(Vehicle):
    def start(self):
        print("Bike starts using a self-start button")


class ElectricCar(Vehicle):
    def start(self):
        print("Electric car starts using a power button")


car = Car()
bike = Bike()
electric = ElectricCar()

car.start()
bike.start()
electric.start()
```
