```python
import mysql.connector

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password="1234"
)

if connection.is_connected():
    print("MySQL connected successfully!")
```

```python
cursor = connection.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS company_db")

print("Database created successfully!")
```

```python
cursor.execute("USE company_db")

cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(100),
    salary INT,
    department VARCHAR(100)
)
""")

print("Employee table created successfully!")
```

```python
sql = """
INSERT INTO employees (name, salary, department)
VALUES (%s, %s, %s)
"""

values = ("Gaurav", 30000, "IT")

cursor.execute(sql, values)
connection.commit()

print("Employee inserted successfully!")
```

```python
cursor.execute("SELECT * FROM employees")

records = cursor.fetchall()

for record in records:
    print(record)
```

```python
sql = "UPDATE employees SET salary = %s WHERE name = %s"

values = (40000, "Gaurav")

cursor.execute(sql, values)
connection.commit()

print("Employee salary updated!")
```

```python
sql = "DELETE FROM employees WHERE name = %s"

values = ("Gaurav",)

cursor.execute(sql, values)
connection.commit()

print("Employee deleted successfully!")
```

```python
cursor.close()
connection.close()

print("MySQL connection closed!")
```
