```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary

employee1 = Employee("Rahul",250000)
employee2 = Employee("priya",30000)
employee3 = Employee("Gaurav",15000)

print(employee1.name , employee1.salary)
print(employee2.name , employee2.salary)
print(employee3.name , employee3.salary)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(self.name, self.salary)


employee1 = Employee("Rahul", 25000)
employee2 = Employee("Priya", 30000)
employee3 = Employee("Gaurav", 15000)

employee1.display()
employee2.display()
employee3.display()
```

```python
class Laptop:
    def __init__(self,brand,price):
        self.brand = brand
        self.price = price

    def show_details(self):
        print(self.brand , self.price)
        
laptop1 = Laptop("Dell",50000)
laptop2 = Laptop("HP",45000)

laptop1.show_details()
laptop2.show_details()
```

```python
class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        print(self.length * self.width)
        
rectangle1 = Rectangle(10,5)


rectangle1.area()
```

```python
class Car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
        
    def show_car(self):
        print(self.brand ,  self.model)

car1 = Car("Toyota","Fortuner")

car1.show_car()
```

```python
# real word example

class Student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def check_result(self):
        if self.marks >= 40:
             print(self.name, "Pass")
        else:
             print(self.name, "Fail")

student1 = Student("Gaurav",75)

student1.check_result()
```
