```python
class Product:
    def __init__(self, name):
        self.name = name

product1 = Product("Laptop")
product2 = product1

print(product1.name)
print(product2.name)

print(product1 is product2)
```

```python
class Employee:
    def __init__(self, name):
        self.name = name

employee1 = Employee("Gaurav")
employee2 = employee1

del employee1

print(employee2.name)
```

```python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Gaurav")
student2 = Student("Gaurav")

print(student1 is student2)
print(student1 == student2)
print(student1.name == student2.name)
```

```python
import gc

print("Garbage collection started")

collected = gc.collect()

print("Objects collected:", collected)
```

```python
import sys

class Product:
    pass
product1 = Product()
print("Reference count:", sys.getrefcount(product1))

product2 = product1
print("Reference count:", sys.getrefcount(product1))

del product2
print("Reference count:", sys.getrefcount(product1))
```
