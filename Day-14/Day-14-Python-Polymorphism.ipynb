```python
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat says Meow")

class Cow:
    def sound(self):
        print("Cow says Moo")

dog1 = Dog()
cat1 = Cat()
cow1 = Cow()

dog1.sound()
cat1.sound()
cow1.sound()
```

```python
class Shape:
    def area(self):
        pass

class Reactangle(Shape):
    def area(self):
        length = 10
        width = 5
        print("Reactangle Area",length * width)

class Square(Shape):
    def area(self):
        side = 4
        print("Square area",side * side)

reactangle1 = Reactangle()
square1 = Square()

reactangle1.area()
square1.area()
```

```python
class Payment:
    def pay(self):
        print("Processing Payment")

class UPIPayment(Payment):
    def pay(self):
        print("Payment done using UPI")

class CardPayment(Payment):
    def pay(self):
        print("Payment done using Card")

upipayment1 = UPIPayment()
cardpayment1 = CardPayment()

upipayment1.pay()
cardpayment1.pay()
```

```python
class Employee:
    def calculate_salary(self):
        print("Calculating Employee Salary")


class FullTimeEmployee(Employee):
    def calculate_salary(self):
        print("Full Time Salary: 30000")


class PartTimeEmployee(Employee):
    def calculate_salary(self):
        print("Part Time Salary: 15000")


employee1 = FullTimeEmployee()
employee2 = PartTimeEmployee()

employee1.calculate_salary()
employee2.calculate_salary()
```
