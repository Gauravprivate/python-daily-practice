```python
numbers = [i * i for i in range(1,11)]

print(numbers)
```

```python
numbers = [i for i in range(1,21) if i % 2 == 1]

print(numbers)
```

```python
numbers = [1, 2, 3, 4, 5]
result = [i * 10 for i in numbers]

print(result)
```

```python
numbers = [1, 2, 3, 4, 5, 6]
result = ["Even" if i % 2 == 0 else "Odd" for i in numbers]

print(result)
```

```python
words = ["Python", "is", "easy", "and", "powerful"]

result = [word for word in words if len(word) > 3]

print(result)
```

```python
numbers = [i for i in range(1, 21)]

result = [i * i for i in numbers if i % 2 == 0]

print(result)
```
